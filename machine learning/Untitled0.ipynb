```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
X,y=make_regression(n_samples=100,n_features=1,noise=10)
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

# create model
lr = LinearRegression()

# fit model
lr.fit(X, y)

# print coefficients
print(lr.coef_)
print(lr.intercept_)

# predict values (this gives points on the line)
y_pred = lr.predict(X)

# plot original data
plt.scatter(X, y, color='blue')

# plot regression line
plt.plot(X, y_pred, color='red')

plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Line")
plt.show()
```

```python
 import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge

# Linear Regression (alpha = 0 equivalent)
lr = LinearRegression()
lr.fit(X, y)

# Ridge models
rr = Ridge(alpha=10)
rr.fit(X, y)

rr1 = Ridge(alpha=100)
rr1.fit(X, y)

# sort X for smooth lines
X_sorted = np.sort(X, axis=0)

# plot points
plt.scatter(X, y, color='blue')

# plot lines
plt.plot(X_sorted, lr.predict(X_sorted), color='red', label='alpha=0')
plt.plot(X_sorted, rr.predict(X_sorted), color='green', label='alpha=10')
plt.plot(X_sorted, rr1.predict(X_sorted), color='orange', label='alpha=100')

plt.legend()
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear vs Ridge Regression")
plt.show()
```
